```python
import os
import chardet

ruta_originales = '../../Datos/Datos originales/'

datasets = {
    "2019": ("GRD_PUBLICO_2019.txt", "CIP_ENCRIPTADO"),
    "2020": ("GRD_PUBLICO_2020.txt", "CIP_ENCRIPTADO"),
    "2021": ("GRD_PUBLICO_2021.txt", "CIP_ENCRIPTADO"),
    "2022": ("GRD_PUBLICO_2022.txt", "CIP_ENCRIPTADO"),
    "2023": ("GRD_PUBLICO_2023.txt", "CIP_ENCRIPTADO"),
    "2024": ("GRD_PUBLICO_2024.txt", "ID_BENEFICIARIO"),
}

for año, (archivo, campo) in datasets.items():
    ruta = os.path.join(ruta_originales, archivo)

    try:
        with open(ruta, "rb") as f:
            data = f.read()  # lee el archivo completo

        resultado = chardet.detect(data)
        cod_detectada = resultado["encoding"]
        confianza = resultado["confidence"]

        print(f"Año {año} (archivo {archivo}):")
        print(f"  Tamaño: {len(data):,} bytes")
        print(f"  Codificación detectada:   {cod_detectada} (confianza {confianza:.2f})")
        print()

    except FileNotFoundError:
        print(f"Año {año}: Archivo {archivo} no encontrado.\n")
```

```python
import os

carpeta = "../../Datos/Datos originales"

archivos = {
    "2019": ("GRD_PUBLICO_2019.txt", "utf-8"),
    "2020": ("GRD_PUBLICO_2020.txt", "utf-8"),
    "2021": ("GRD_PUBLICO_2021.txt", "utf-8-sig"),
    "2022": ("GRD_PUBLICO_2022.txt", "utf-16"),
    "2023": ("GRD_PUBLICO_2023.txt", "utf-16"),
    "2024": ("GRD_PUBLICO_2024.txt", "latin1")
}

for año, (archivo, encoding) in archivos.items():

    ruta = os.path.join(carpeta, archivo)

    try:
        with open(ruta, "r", encoding=encoding) as f:

            encabezado = next(f)
            columnas_esperadas = len(encabezado.rstrip("\n\r").split("|"))

            total_filas = 0
            filas_problematicas = 0

            ejemplos = []

            for num_linea, linea in enumerate(f, start=2):

                total_filas += 1

                columnas_actuales = len(
                    linea.rstrip("\n\r").split("|")
                )

                if columnas_actuales != columnas_esperadas:

                    filas_problematicas += 1

                    if len(ejemplos) < 10:
                        ejemplos.append(
                            (num_linea,
                             columnas_actuales,
                             columnas_esperadas)
                        )

        print(f"\nAño {año}")
        print(f"Archivo: {archivo}")
        print(f"Columnas esperadas: {columnas_esperadas}")
        print(f"Filas de datos: {total_filas:,}")
        print(f"Filas problemáticas: {filas_problematicas:,}")

        if ejemplos:
            print("Primeros ejemplos:")
            for linea, encontradas, esperadas in ejemplos:
                print(
                    f"  Línea {linea}: "
                    f"{encontradas} columnas "
                    f"(esperadas {esperadas})"
                )

        print("-" * 80)

    except Exception as e:
        print(f"\nAño {año}")
        print(f"Error al procesar {archivo}: {e}")
        print("-" * 80)
```

```python
import os

# ==================== COMPARACIÓN DE ESQUEMAS ====================

# Diccionario para almacenar el conjunto de columnas de cada archivo
columnas_por_archivo = {}

# Leer únicamente la cabecera de cada archivo
for año, (archivo, encoding) in archivos.items():

    ruta = os.path.join(carpeta, archivo)

    with open(ruta, "r", encoding=encoding) as f:
        encabezado = f.readline().strip()

    columnas = encabezado.split("|")

    columnas_por_archivo[archivo] = set(columnas)

# Lista de nombres de archivos para comparar
nombres_archivos = list(columnas_por_archivo.keys())

# ==================== ANÁLISIS COMPARATIVO ====================

for i in range(len(nombres_archivos)):
    for j in range(i + 1, len(nombres_archivos)):

        a1 = nombres_archivos[i]
        a2 = nombres_archivos[j]

        cols1 = columnas_por_archivo[a1]
        cols2 = columnas_por_archivo[a2]

        faltan_en_a1 = cols2 - cols1
        faltan_en_a2 = cols1 - cols2

        if faltan_en_a1 or faltan_en_a2:

            print(f"\nADVERTENCIA: Diferencias entre {a1} y {a2}")

            if faltan_en_a1:
                print(f"  {a1} NO tiene:")
                for col in sorted(faltan_en_a1):
                    print(f"    - {col}")

            if faltan_en_a2:
                print(f"  {a2} NO tiene:")
                for col in sorted(faltan_en_a2):
                    print(f"    - {col}")

        else:
            print(f"✔︎ CORRECTO: {a1} y {a2} tienen las mismas columnas")
```
